# Situação 1:
A escola em que estamos construindo o nosso case de dados compartilhou os dados das notas de um estudante para que pudéssemos calcular a média deste em até uma casa decimal.

Os dados recebidos correspondem a um dicioná rio com as chaves indicando o trimestre em questão e os valores das notas de cada trimestre do estudante em uma dad matéria.


```python
#Nota do(a) estudante

notas = {'1° Trimestre': 8.5, '2° Trimestre': 9.5, '3° Trimestre': 7}
notas
```

```python
# calculando a soma

soma = 0

for nota  in notas.values():
  soma += nota
soma
```

```python
# usando a função embutida sun()

soma = sum(notas.values())
soma
```

```python
# usandp a função embutida len()

qtd_notas = len(notas)
qtd_notas
```

```python
# calculando a média
media = soma / qtd_notas
media
```

```python
# arrendondar a média usando round():
media = soma / qtd_notas
media = round(media,1)
media
```

```python
# notas do(a) estudante
notas = [6.0, 7.0, 9.0, 5.0]
```

```python
def boletim(lista):
  media = sum(lista) / len(lista)

  if media >= 6:
    situacao = 'Aprovado'
  else:
    situacao = 'Reprovado'

  return (media, situacao)
```

```python
boletim(notas)
```

```python
media, situacao = boletim(notas)
```

```python
media
```

```python
situacao
```

```python
print(f'O(a) estudante atigiu uma média de {media} e foi {situacao}.')
```

# Função Lambda


```python
#Comparando uma função de qualitativo no formato de função para função anônima
nota = float(input("Digite a nota do(a) estudante: "))

def qualitativo(x):
  return x + 0.5

qualitativo(nota)
```

```python
#Comparando uma função de qualitativo no formato de função para uma função lambda
nota = float(input("Digite a nota do(a) estudante: "))

qualitativo = lambda x: x + 0.5

qualitativo(nota)
```

```python
#Recebendo as notas e calculando a média ponderavél

N1 = float(input("Digite a nota 1: "))
N2 = float(input("Digite a nota 2: "))
N3 = float(input("Digite a nota 3: "))

media_ponderavel = lambda X, Y, Z: (X * 3 + Y * 2 + Z * 5)/10
media_estudante = media_ponderavel(N1, N2, N3)
media_estudante
```

```python
# Exibindo a média
print(f'O(a) estudante atingiu uma média de {media_estudante}')
```

# Mapeamento de valores

```python
# Notas do(a) estudante
notas = [6.0, 7.0, 9.0, 5.5, 8.0]
qualitativo = 0.5
```

```python
#Não conseguimos aplicar o lambda em listas direto, é necessário, utilizar junto a ela a função map.
notas_atualizadas = map (lambda x: x + qualitativo, notas)
notas_atualizadas
```

```python
notas_atualizadas = list(notas_atualizadas)
notas_atualizadas
```
